```python
import pandas as pd
from pandasql import sqldf
```

```python
df = pd.read_csv("data.csv")
def q(query):
    return sqldf(query, globals())
```

```python
q("""SELECT bot,
       COUNT(*) AS sessions,
       ROUND(AVG(clicks), 1)                   AS avg_clicks,
       ROUND(AVG(median_click_interval_ms), 1) AS avg_median_interval_ms,
       ROUND(AVG(click_interval_std_ms), 1)    AS avg_interval_std_ms,
       ROUND(AVG(button_hover_ms), 1)          AS avg_button_hover_ms
FROM df
GROUP BY bot""")
```
