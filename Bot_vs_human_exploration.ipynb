```python
import pandas as pd
from pandasql import sqldf
```

```python
df = pd.read_csv("data.csv")
def q(query):
    return sqldf(query, globals())
```

```python
q("""SELECT bot,
       COUNT(*) AS sessions,
       ROUND(AVG(clicks), 1)                   AS avg_clicks,
       ROUND(AVG(median_click_interval_ms), 1) AS avg_median_interval_ms,
       ROUND(AVG(click_interval_std_ms), 1)    AS avg_interval_std_ms,
       ROUND(AVG(button_hover_ms), 1)          AS avg_button_hover_ms
FROM df
GROUP BY bot""")
# Compare click behaviour of humans (bot = 0) and bots (bot = 1)
```

```python
q("""SELECT referrer_type,
       SUM(bot)            AS bots,
       COUNT(*) - SUM(bot) AS humans
FROM df
GROUP BY referrer_type
ORDER BY bots DESC""")

# Count bots and humans per referrer type which shows that the ratio is similar
# so the reffere_type collum is not a good metric to measure(bot or a human).
```

```python
# Mouse and scrolling behaviour per group (bot = 0: human, bot = 1: bot)
q("""SELECT bot,
       COUNT(*) AS sessions,
       ROUND(AVG(mouse_path_curvature), 3) AS avg_mouse_curvature,
       ROUND(AVG(idle_ratio), 3)           AS avg_idle_ratio,
       ROUND(AVG(scroll_entropy), 3)       AS avg_scroll_entropy
FROM df
GROUP BY bot""")
```

Data Summary and Cleaning Report (Group 48)

The dataset (08_bot_vs_human.csv) holds 400 web shop sessions recorded by the shop’s anti-abuse team with one row per session. 

Features

Mouse Behaviour: clicks, intervals between clicks, path curvature, path distance
Keyboard Behaviour: digraph latency, dwell time on three form fields
Activity Patterns: idle ratio, scroll entropy, copy events, button hover interval
Session Metadata: session duration, pages visited, timezone mismatch, referrer type, viewport width, webdriver flag
Bot: 1 = confirmed automated / 0 = human

Cleaning Steps Applied

Removed the “referrer_type” column as it has little to no impact on the identification of a bot.
Removed the records with their “digraph_latency_ms” below 80 ms AND are considered human (Bot = 0) because they can shift the predictions as they are extreme outliers.
There werent any missing values in the dataset.
